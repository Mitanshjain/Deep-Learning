# ANN (Artificial Neural Network)

```python
# So when we talking about the deep learning there is term like Artificial Neural Network so what is the Artificial Neural Network(ANN)?
# So ANN is stands for Artificial Neural Network, our human brain contains billion of neurons.These neurons communicate with each other
# and process information.

# So this Artificial Neural Network are inspired from human neuron system.

# Similarly, an artificial Neural Network contains artificial neurons(also called nodes).

# Each neuron receive information, perform some mathematical calculations, and passes the result forward.
```

```python
# So the ANN generally contains three layers :-
# (1).Input layer :- This is where we can provide data to our neural network.for example :- House size, No. of bedrooms...etc
# (2).Hidden layer :- This is where actual learning and mathematical processing happens. for example :- Weight, biases, Activation function.
# (3).Output layer :- This produce final prediction.for example :- House price = 50 lakh
```

```python
# Loss function in DL:-
# loss_function :- Difference between Actual - Prediction.

# When we have target column is numerical
# Loss ---> MSE(Mean Squared Error)

# When we have target column is categorical
# Loss ---> 
# (1). When we have 2 sub-category ---> then loss ---> Binary_cross_entropy
# (2). When we have more than 2 sub-category ---> then loss ---> Sparse_categorical_cross_entropy
```

```python
# Backward propagation ---> Optimization technique
# dy/dx11=(dy/dx31*dx31/dx21*dx21/dx11)+(dy/dx31 * dx31/dx22 * dx22/dx11)
# dy/dx12=(dy/dx31* dx31/dx21* dx21/dx12) +(dy/dx31 * dx31/dx22 * dx22/dx12)
# dy/dx13=(dy/dx31*dx31/dx21* dx21/dx13) +(dy/dx31 * dx31/dx22 * dx22/dx13)
```

```python
So we have two type of propagation :-
(1).Forward propagation :- Forward propagation is the process in which output data travel from the input layer through the hidden layer
to the output layer to generate prediction.

(2).Backward propagation :- Backward propagation is the process of calculating how much each weight contributed to the prediction error
and determining how those weight should change to reduce the error.
```

```python
(1).Forward propagation :- Making a prediction and direction is forward.
(2). Backward propagation :- learning from the prediction error and direction is backward.
```
