```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("Attrition.csv")
```

```python
df.head(4)
```

```python
import tensorflow as tf
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder,StandardScaler
from sklearn.metrics import accuracy_score
```

```python
df.columns
```

```python
df = df.dropna()
```

```python
x = df.drop(columns = ['Attrition'])
y = df['Attrition']
```

```python
x.info()
```

```python
categorical_features = ['BusinessTravel',
                        'Department',
                        'EducationField',
                        'Gender',
                        'JobRole',
                        'MaritalStatus',
                        'Over18',
                        'OverTime']
```

```python
x_encoded = pd.get_dummies(x,columns=categorical_features,drop_first=True,dtype=int)
```

```python
x_encoded = x_encoded.astype(int)
```

```python
x_encoded
```

```python
# Encode the target data
lb = LabelEncoder()
```

```python
y_enocded = lb.fit_transform(y)
```

```python
y_enocded  # It gives array
```

```python
x_train,x_test,y_train,y_test = train_test_split(x_encoded,y_enocded,test_size=0.2,random_state=42)
```

```python
x_train.head(3)
```

```python
sc = StandardScaler()
```

```python
x_train_scaled = sc.fit_transform(x_train)
x_test_scaled = sc.transform(x_test)
```

# Now we will create Architecture of our model

```python
model = tf.keras.models.Sequential([
    tf.keras.layers.Dense(units=128, activation='relu',
                         input_dim = x_train_scaled.shape[1]),  # This is the input layer and choose shape is 1 for the columns if we want to choose rows then we select 0.

    tf.keras.layers.Dropout(0.3), # Dropout randomly turns off 30% of neurons during each training step.
    # For example, suppose you have 100 neurons in a hidden layer:-
    # 30 neurons will be randomly deactivated.
    # 70 neurons will remain active.
    # The randomly deactivated neurons change during training.

    # We do dropout To prevent Overfitting your model memorize training data well and it perform well on training data but it fails on testing data.


    tf.keras.layers.Dense(units = 64,activation='relu'),  # Hidden layer
    tf.keras.layers.Dense(units=1,activation='sigmoid') # Output layer
])
```

# Compile the model

```python
loss_fn = tf.keras.losses.BinaryCrossentropy() # we choose Binary cross entropy because our target column is in yes or no.
metrics = ['accuracy']
```

```python
learning_rate = 0.001
momentum = 0.9
optimizer = tf.keras.optimizers.SGD(learning_rate=learning_rate,
                                    momentum = momentum, nesterov = True)
```

```python
model.compile(optimizer=optimizer, loss=loss_fn,metrics=metrics)
```

# Train the model

```python
model.fit(x_train_scaled,
          y_train,
          epochs = 50,
          batch_size = 8,
          validation_split = 0.1) # This is our those data we are used for testing.
```

Epochs :- An Epochs is one complete pass of the entire training dataset through a neural network. During each epoch the model learns by updating its weight through forward propagation and backward propagation.

Batch_size :- Batches is the smaller chunk of the data.

```python
loss , accuracy = model.evaluate(x_test_scaled, y_test)
```

```python
print(f"test loss = {loss:.4f}, test accuracy = {accuracy:.4f}")
```
